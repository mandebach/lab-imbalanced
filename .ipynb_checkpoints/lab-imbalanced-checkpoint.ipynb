# LAB | Imbalanced

**Load the data**

In this challenge, we will be working with Credit Card Fraud dataset.

https://raw.githubusercontent.com/data-bootcamp-v4/data/main/card_transdata.csv

Metadata

- **distance_from_home:** the distance from home where the transaction happened.
- **distance_from_last_transaction:** the distance from last transaction happened.
- **ratio_to_median_purchase_price:** Ratio of purchased price transaction to median purchase price.
- **repeat_retailer:** Is the transaction happened from same retailer.
- **used_chip:** Is the transaction through chip (credit card).
- **used_pin_number:** Is the transaction happened by using PIN number.
- **online_order:** Is the transaction an online order.
- **fraud:** Is the transaction fraudulent. **0=legit** -  **1=fraud**


```python
#Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import precision_score, recall_score, classification_report, confusion_matrix, f1_score
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDClassifier
from imblearn.under_sampling import RandomUnderSampler
from imblearn.over_sampling import RandomOverSampler, SMOTE
from sklearn.utils import resample

SEED=1
```

```python
fraud = pd.read_csv("https://raw.githubusercontent.com/data-bootcamp-v4/data/main/card_transdata.csv")
fraud.head()
```

**Steps:**

- **1.** What is the distribution of our target variable? Can we say we're dealing with an imbalanced dataset?
- **2.** Train a LogisticRegression.
- **3.** Evaluate your model. Take in consideration class importance, and evaluate it by selection the correct metric.
- **4.** Run **Oversample** in order to balance our target variable and repeat the steps above, now with balanced data. Does it improve the performance of our model? 
- **5.** Now, run **Undersample** in order to balance our target variable and repeat the steps above (1-3), now with balanced data. Does it improve the performance of our model?
- **6.** Finally, run **SMOTE** in order to balance our target variable and repeat the steps above (1-3), now with balanced data. Does it improve the performance of our model? 

```python
fraud.describe()
```

```python
# 1. What is the distribution of our target variable? Can we say we're dealing with an imbalanced dataset?

class_counts = fraud['fraud'].value_counts()
print(class_counts)
print()

# A model predicting "No Fraud" for every single transaction would already score:
naive_accuracy = class_counts[0] / len(fraud)
print(f"Accuracy (always predicting 'No Fraud'): {naive_accuracy * 100:.1f}%")
```

```python
# Yes we're dealing with an imbalanced dataset because there are far more non-fraud cases than fraud cases.
class_counts.plot(kind="bar")
plt.show()
```

```python
fraud.info()
```

```python
# 2. Train a LogisticRegression.

# Separate features (X) and target (y)
X = fraud.drop(columns=['fraud'])  # Everything except the label
y = fraud['fraud']
```

```python
# Train Test Split

X_train, X_test, y_train, y_test = train_test_split(
    X,y,
    test_size=0.2,
    random_state=SEED,
    stratify=y)

print(f'X_train shape: {X_train.shape}')
print(f'X_test shape:  {X_test.shape}')
```

```python
# Confirm stratify worked --> same imbalance ratio preserved in both Train and Test
print('Train:', y_train.value_counts(normalize=True))
print('Test: ', y_test.value_counts(normalize=True))
```

```python
# Feature Scaling

scaler = StandardScaler()
scaler.set_output(transform="pandas")

X_train_scaled_df = scaler.fit_transform(X_train)
X_test_scaled_df = scaler.transform(X_test)
```

```python
# BEFORE: original scale
print('BEFORE scaling (X_train):')
print(X_train.describe().loc[['mean', 'std']].round(4))
print()

# AFTER: standardized scale
print('AFTER scaling (X_train):')
print(X_train_scaled_df.describe().loc[['mean', 'std']].round(4))
```

```python
# Model Training

baseline_model = SGDClassifier(loss="log_loss", random_state=SEED)
baseline_model.fit(X_train_scaled_df, y_train)  # Fit on the original, imbalanced set

print("Model trained!")
```

```python
# 3. Evaluate your model. Take in consideration class importance, and evaluate it by selection the correct metric.
# Results

# Evaluate the model on both Train and Test, to check for overfitting
y_pred_train = baseline_model.predict(X_train_scaled_df)     # train
y_pred_test = baseline_model.predict(X_test_scaled_df)       # test

accuracy_train = accuracy_score(y_train, y_pred_train)  # train, just to compare with the report below
print(f'Accuracy (Train): {accuracy_train * 100:.2f}%')
print()

# Classification Report (Test)
# ----------------------------
print(classification_report(y_test, y_pred_test, target_names=['No Fraud', 'Fraud'], zero_division=0))
```

```python
# Analysis: The accuracy on Train and Test is very good at 96%. No underfitting.
# But the recall on Fraud is very low at 61%

# Fraud Precision: 89
# Fraud Recall: 61 (It caught only 61% of actual fraud cases)
# Fraud F1 Score: 72
# Overall Accuracy: 96
```

```python
# 4. Run Oversample in order to balance our target variable and repeat the steps above, now with balanced data. 
# Does it improve the performance of our model? YES.
```

```python
# RandomOverSampler --> duplicates random instances from the minority class (Fraud)
# until it matches the size of the majority class (No Fraud)
ros = RandomOverSampler(random_state=SEED)
X_train_over, y_train_over = ros.fit_resample(X_train_scaled_df, y_train)

print('Before:', y_train.value_counts())
print('After: ', y_train_over.value_counts())

oversampling_model = SGDClassifier(loss='log_loss', random_state=SEED)
oversampling_model.fit(X_train_over, y_train_over)  # Fit on the oversampled Train set

print("Model trained!")
```

```python
# Evaluate the model on both Train and Test, to check for overfitting
y_pred_train = oversampling_model.predict(X_train_over)     # train (oversampled)
y_pred_test = oversampling_model.predict(X_test_scaled_df)  # test (untouched!)

accuracy_train = accuracy_score(y_train_over, y_pred_train)
print(f'Accuracy (Train): {accuracy_train * 100:.2f}%')
print()

# Classification Report (Test)
# ----------------------------
print(classification_report(y_test, y_pred_test, target_names=['No Fraud', 'Fraud'], zero_division=0))
```

```python
# Analysis: The accuracy has dropped slightly to 93% but the recall has improved dramatically, up from 61% to 95%. 
# However the precision on Fraud has dropped from 89% to 57%. 

# Fraud Precision: 57 - This means roughly 43% of flagged transactions are now false alarms (false positives).
# Fraud Recall: 95 
# Fraud F1 Score: 71
# Overall Accuracy: 93
```

```python
# 5. Now, run Undersample in order to balance our target variable and repeat the steps above (1-3), now with balanced data. 
# Does it improve the performance of our model?
```

```python
rus = RandomUnderSampler(random_state=SEED)
X_train_under, y_train_under = rus.fit_resample(X_train_scaled_df, y_train)

print('Before:', y_train.value_counts())
print('After: ', y_train_under.value_counts())

undersampling_model = SGDClassifier(loss='log_loss', random_state=SEED)
undersampling_model.fit(X_train_under, y_train_under)  # Fit on the undersampled Train set

print("Model trained!")
```

```python
y_pred_train = undersampling_model.predict(X_train_under)   # train (undersampled)
y_pred_test = undersampling_model.predict(X_test_scaled_df) # test (untouched!)

accuracy_train = accuracy_score(y_train_under, y_pred_train)
print(f'Accuracy (Train): {accuracy_train * 100:.2f}%')
print()

# Classification Report (Test)
# ----------------------------
print(classification_report(y_test, y_pred_test, target_names=['No Fraud', 'Fraud'], zero_division=0))
```

```python
# Analysis: Accuracy and Recall improve slightly. Precision on Fraud has dropped to 55%.

# Fraud Precision: 55
# Fraud Recall: 96
# Fraud F1 Score: 70
# Overall Accuracy: 93
```

```python
# 6. Finally, run SMOTE in order to balance our target variable and repeat the steps above (1-3), now with balanced data. 
# Does it improve the performance of our model?
```

```python
sm = SMOTE(random_state=SEED)
X_train_smote, y_train_smote = sm.fit_resample(X_train_scaled_df, y_train)

print('Before:', y_train.value_counts())
print('After: ', y_train_smote.value_counts())

smote_model = SGDClassifier(loss='log_loss', random_state=SEED)
smote_model.fit(X_train_smote, y_train_smote)  # Fit on the SMOTE-resampled Train set

print("Model trained!")
```

```python
y_pred_train = smote_model.predict(X_train_smote)
y_pred_test = smote_model.predict(X_test_scaled_df)

accuracy_train = accuracy_score(y_train_smote, y_pred_train)
print(f'Accuracy (Train): {accuracy_train * 100:.2f}%')
print()

print(classification_report(y_test, y_pred_test, target_names=['No Fraud', 'Fraud'], zero_division=0))
```

```python
# Fraud Precision: 57
# Fraud Recall: 95
# Fraud F1 Score: 71 
# Overall Accuracy: 93
```
